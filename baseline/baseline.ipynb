```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set_theme(style='darkgrid', palette='Set2', context='talk')
```

```python
# Loss by epoch
run1_loss = list(pd.read_csv('run1_loss.csv')['prosqa-coconut - eval/loss']) # epochs 1, 2
run2_loss = list(pd.read_csv('run2_loss.csv')['prosqa-coconut - eval/loss']) # epochs 3-

losses = run1_loss + run2_loss
epochs = range(1, len(losses) + 1)


fig, ax = plt.subplots(figsize=plt.figaspect(0.6), dpi=100, constrained_layout=True)
sns.lineplot(y=losses, x=epochs, ax=ax, marker='o')
ax.set_title(f'ProsQA Training Loss')
ax.set_xlabel('Epoch')
plt.xlim((0, 31))
plt.savefig('prosqa_train_loss.png')
plt.show()
```

```python
# Accuracy by stage/epoch
acc = list(pd.read_csv('run2_acc.csv')['prosqa-coconut - eval/acc']) # epochs 5, 10, ...
stages = range(6)

fig, ax = plt.subplots(figsize=plt.figaspect(0.6), dpi=100, constrained_layout=True)
sns.lineplot(y=[a * 100 for a in acc], x=stages, ax=ax, marker='o')
ax.set_title(f'ProsQA Validation Accuracy')
ax.set_xlabel('Stage ($k$)')
ax.set_ylabel('%')
plt.savefig('prosqa_val_acc.png')
plt.show()
```
